# Figures and tables of the thesis
*Multi-Objective Optimisation of Agrivoltaic System Design: Balancing Land Equivalent Ratio and Revenue* (Hao Li, 2026)

This notebook regenerates the figures of the thesis from the stored results with the scripts in `analysis/` (Annex A3, Table 17). No ray tracing or crop simulation is run; each step takes a few seconds.

**Requirements:** the `agripv` environment as kernel and the stored results in `results/` (on a fresh clone: download `thesis_data.zip` from the release v1.0-thesis and extract it into this folder, see README.md).

Run the cells from top to bottom (*Kernel → Restart & Run All*).

```python
import os
import sys
from pathlib import Path
if "agripv" not in sys.executable:  # other Python installations have other package versions
    raise RuntimeError(f"Wrong kernel ({sys.executable}): choose Kernel -> Change kernel -> Python (agripv)")
from IPython.display import Image, display

ROOT = Path.cwd()
while not (ROOT/"analysis"/"reproduce_thesis.py").exists():  # find the repository folder
    if ROOT.parent == ROOT:
        raise FileNotFoundError("Start this notebook inside the optimizer repository.")
    ROOT = ROOT.parent
os.chdir(ROOT)
RESULTS = ROOT/"results"
assert (RESULTS/"recheck_combined_192skies").exists(), "results/ missing: extract thesis_data.zip first (README.md)"

def show(path, width=850):
    display(Image(filename=str(RESULTS/path), width=width))

print("Repository:", ROOT)
```
